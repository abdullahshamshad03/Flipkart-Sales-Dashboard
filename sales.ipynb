```python
towns_to_greater_accra = ["Dawhenya", "Dodowa", "Amasaman", "Gbawe", "Oyarifa", "Prampram", "Weija"]

def clean_region(loc):
    if loc in towns_to_greater_accra:
        return "Greater Accra"
    elif loc == "Kasoa":
        return "Central"
    elif loc == "Kitase":
        return "Eastern"
    elif loc == "Brong-Ahafo":
        return "Bono"
    return loc

df_copy["Region"] = df_copy["Location"].apply(clean_region)
    
```

```python
df_copy.groupby('Region')['Delivery Days'].mean().sort_values(ascending=True).plot(kind= 'barh')
```
